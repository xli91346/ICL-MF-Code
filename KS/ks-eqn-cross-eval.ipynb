```python
# ============================================================
# KAGGLE ONE-CELL
# KS BETA-OPERATOR / 20-TRAJECTORY
# STRICT PAIRED GROUP-B OOD EVALUATION
# ============================================================
#
# DATASET ROOT
# ------------
#   /kaggle/input/notebooks/xinyuanli913/ks-dataset
#
# NO-MIXED ROOT
# -------------
#   /kaggle/input/notebooks/xinyuanli913/ks-eqn-no-mixed
#
# MIXED ROOT
# ----------
#   /kaggle/input/notebooks/xinyuanli913/ks-eqn-mixed
#
# LOCKED DATASET / TRAINING SEMANTICS
# -----------------------------------
#   operator = beta
#   200 beta operators in [0.070,0.085]
#   20 independent fixed-IC trajectories / beta
#   uc/uf shape = [200,20,150,100]
#
#   BASE = 2.505, 2.510, ..., 3.000  (100 tokens)
#   OOD  = 3.005, 3.010, ..., 3.250  (50 tokens)
#
#   Group B = first 100 entries of seed-42 permutation
#   Group A = remaining 100 entries
#
# FORMAL STRICT PAIRED EVALUATION
# -------------------------------
#   - BEST train-monitor checkpoint from NO-MIXED
#   - BEST train-monitor checkpoint from MIXED
#   - Group B only
#   - all 100 beta operators
#   - all 20 trajectories/operator
#   - all 50 OOD query times
#
#   total query cases:
#       100 * 20 * 50 = 100,000
#
#   M = 5 demonstrations per case.
#
# STRICT DEMO RULE
# ----------------
#   For each Group-B (operator,trajectory), define
#
#       D_NO(op,traj)
#           = BASE times with NO-MIXED BEST usage > 0
#
#       D_MIX(op,traj)
#           = BASE times with MIXED BEST usage > 0
#
#       D_SHARED(op,traj)
#           = D_NO intersection D_MIX
#
#   The M=5 formal demos are selected ONLY from D_SHARED.
#   Therefore every evaluation demo was actually consumed as a
#   gradient-training demonstration by BOTH selected BEST models.
#
#   Query/target appearances and monitor appearances are not counted
#   by the training scripts and therefore do not qualify.
#
# PAIRING
# -------
#   NO-MIXED and MIXED use exactly the same:
#       operator
#       trajectory
#       5 demo times
#       OOD query time
#       LF query
#       HF target
#
#   RAW LF, NO-MIXED and MIXED are compared case-by-case.
#
# IMPORTANT
# ---------
#   Group-B OOD was never used in gradient training, normalization,
#   monitor construction, or checkpoint selection.
# ============================================================

from __future__ import annotations

import json
import time
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
import torch.nn as nn


# ============================================================
# 0. CONFIG
# ============================================================

DATASET_ROOT = Path(
    "/kaggle/input/notebooks/xinyuanli913/ks-dataset"
)

NOMIXED_ROOT = Path(
    "/kaggle/input/notebooks/xinyuanli913/ks-eqn-no-mixed"
)

MIXED_ROOT = Path(
    "/kaggle/input/notebooks/xinyuanli913/ks-eqn-mixed"
)

EXPECTED_DATASET_NAME = (
    "ks_mf_queryt_beta070_085_"
    "FD2LF72_SpecHF256_T3p25_"
    "B100late_OOD50_dt001_Nop200_Ntraj20_Nt150_Nr100.npz"
)

OUT_DIR = Path(
    "/kaggle/working/"
    "ks_beta20_latebase_B100_OOD50_T3p25_"
    "strictSharedDemo_paired_eval_all100000"
)
OUT_DIR.mkdir(parents=True, exist_ok=True)

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

EVAL_BATCH_SIZE = 512 if DEVICE == "cuda" else 128
EVAL_SEED = 424242
FIXED_M = 5
MAX_CTX_ROWS = 23
EPS = 1.0e-8

SAVE_PREDICTIONS = True

N_OP_EXPECTED = 200
N_TRAJ_EXPECTED = 20
N_TIME_EXPECTED = 150
NR_EXPECTED = 100
N_GROUP_EXPECTED = 100
N_BASE_EXPECTED = 100
N_OOD_EXPECTED = 50

BASE_MIN_EXPECTED = 2.505
BASE_MAX_EXPECTED = 3.000
OOD_MIN_EXPECTED = 3.005
OOD_MAX_EXPECTED = 3.250

SOLVER_DT_EXPECTED = 0.001
SNAPSHOT_DT_EXPECTED = 0.005

PREFERRED_ARCH = {
    "D_MODEL": 288,
    "N_HEAD": 6,
    "N_LAYERS": 5,
    "D_FF": 576,
    "DROPOUT": 0.10,
    "N_PARAMS": 3_484_036,
}

TYPE_X_DEMO = 0
TYPE_T_DEMO = 1
TYPE_UC_DEMO = 2
TYPE_UF_DEMO = 3
TYPE_X_QUERY = 4
TYPE_T_QUERY = 5
TYPE_UC_QUERY = 6
N_ROW_TYPES = 7

if DEVICE == "cuda":
    torch.backends.cuda.matmul.allow_tf32 = True
    torch.backends.cudnn.allow_tf32 = True

print("=" * 136)
print("KS BETA20 STRICT-SHARED-DEMO PAIRED GROUP-B OOD EVALUATION")
print("=" * 136)
print("DEVICE       :", DEVICE)
print("DATASET_ROOT :", DATASET_ROOT)
print("NOMIXED_ROOT :", NOMIXED_ROOT)
print("MIXED_ROOT   :", MIXED_ROOT)
print("OUT_DIR      :", OUT_DIR)
print("Protocol     : BEST checkpoints | strict shared-used BASE demos | M=5")
print("Coverage     : 100 Group-B beta x 20 trajectories x 50 OOD times")
print("Expected     : 100,000 exact paired query cases")
print("=" * 136)


# ============================================================
# 1. GENERIC HELPERS
# ============================================================

def torch_load_any(path, map_location="cpu"):
    try:
        return torch.load(
            path,
            map_location=map_location,
            weights_only=False,
        )
    except TypeError:
        return torch.load(
            path,
            map_location=map_location,
        )


def scalar_str(x):
    arr = np.asarray(x)
    value = arr.item() if arr.size == 1 else arr.reshape(-1)[0]
    if isinstance(value, bytes):
        value = value.decode("utf-8")
    return str(value)


def summarize(arr, scale=1.0):
    arr = np.asarray(arr, dtype=np.float64) * scale
    return {
        "min": float(np.min(arr)),
        "max": float(np.max(arr)),
        "mean": float(np.mean(arr)),
        "median": float(np.median(arr)),
        "std": float(np.std(arr)),
        "p90": float(np.quantile(arr, 0.90)),
        "p95": float(np.quantile(arr, 0.95)),
        "p99": float(np.quantile(arr, 0.99)),
    }


def summary_str(s):
    return (
        f"mean={s['mean']:.4f}% | "
        f"median={s['median']:.4f}% | "
        f"std={s['std']:.4f}% | "
        f"p90={s['p90']:.4f}% | "
        f"p95={s['p95']:.4f}% | "
        f"max={s['max']:.4f}%"
    )


def rel_l2_np(pred, target):
    pred = np.asarray(pred)
    target = np.asarray(target)
    return (
        np.linalg.norm(pred - target, axis=1)
        / np.maximum(
            np.linalg.norm(target, axis=1),
            EPS,
        )
    )


def r2_np(pred, target):
    pred = np.asarray(pred)
    target = np.asarray(target)

    target_mean = np.mean(
        target,
        axis=1,
        keepdims=True,
    )

    sse = np.sum(
        (pred - target) ** 2,
        axis=1,
    )

    sst = np.maximum(
        np.sum(
            (target - target_mean) ** 2,
            axis=1,
        ),
        EPS,
    )

    return 1.0 - sse / sst


# ============================================================
# 2. LOCATE EXACT DATASET
# ============================================================

def locate_dataset(root: Path) -> Path:
    if not root.is_dir():
        raise FileNotFoundError(
            f"Dataset root does not exist:\n{root}"
        )

    exact = sorted(
        root.rglob(EXPECTED_DATASET_NAME)
    )

    if len(exact) == 1:
        return exact[0]

    if len(exact) > 1:
        raise RuntimeError(
            "Multiple exact dataset matches found:\n"
            + "\n".join(str(p) for p in exact)
        )

    # Strict metadata fallback, only if exact filename is absent.
    compatible = []

    for p in sorted(root.rglob("*.npz")):
        try:
            with np.load(p, allow_pickle=False) as f:
                if not {"uc", "uf", "times"}.issubset(f.files):
                    continue

                if tuple(f["uc"].shape) != (
                    N_OP_EXPECTED,
                    N_TRAJ_EXPECTED,
                    N_TIME_EXPECTED,
                    NR_EXPECTED,
                ):
                    continue

                if tuple(f["uf"].shape) != tuple(f["uc"].shape):
                    continue

                ts = np.asarray(
                    f["times"],
                    dtype=np.float64,
                ).reshape(-1)

                if ts.shape != (N_TIME_EXPECTED,):
                    continue

                if not np.isclose(
                    ts[0],
                    BASE_MIN_EXPECTED,
                    atol=1e-12,
                ):
                    continue

                if not np.isclose(
                    ts[-1],
                    OOD_MAX_EXPECTED,
                    atol=1e-12,
                ):
                    continue

                if "base_time_idx" not in f or "ood_time_idx" not in f:
                    continue

                bi = np.asarray(
                    f["base_time_idx"],
                    dtype=np.int64,
                ).reshape(-1)

                oi = np.asarray(
                    f["ood_time_idx"],
                    dtype=np.int64,
                ).reshape(-1)

                if not np.array_equal(
                    bi,
                    np.arange(0, 100, dtype=np.int64),
                ):
                    continue

                if not np.array_equal(
                    oi,
                    np.arange(100, 150, dtype=np.int64),
                ):
                    continue

                compatible.append(p)

        except Exception:
            continue

    if len(compatible) != 1:
        raise RuntimeError(
            "Could not identify exactly one compatible late-base T3p25 dataset.\n"
            "Compatible candidates:\n"
            + "\n".join(str(p) for p in compatible)
        )

    return compatible[0]


DATASET_PATH = locate_dataset(DATASET_ROOT)


# ============================================================
# 3. LOCATE THE CORRECT STRICT-DEMO BEST CHECKPOINTS
# ============================================================

def checkpoint_semantic_reasons(ckpt, expected_mode):
    cfg = ckpt.get("config", {})
    reasons = []

    if cfg.get("TRAIN_MODE") != expected_mode:
        reasons.append(
            f"TRAIN_MODE={cfg.get('TRAIN_MODE')}"
        )

    if cfg.get("operator_parameter") != "beta":
        reasons.append(
            f"operator_parameter={cfg.get('operator_parameter')}"
        )

    if int(cfg.get("N_OP", -1)) != 200:
        reasons.append(
            f"N_OP={cfg.get('N_OP')}"
        )

    if int(cfg.get("N_TRAJ", -1)) != 20:
        reasons.append(
            f"N_TRAJ={cfg.get('N_TRAJ')}"
        )

    if int(cfg.get("N_TIME", -1)) != 150:
        reasons.append(
            f"N_TIME={cfg.get('N_TIME')}"
        )

    if int(cfg.get("NX", -1)) != 100:
        reasons.append(
            f"NX={cfg.get('NX')}"
        )

    if not bool(
        cfg.get("demo_usage_tracking", False)
    ):
        reasons.append("demo_usage_tracking=False/missing")

    if "demo_usage" not in ckpt:
        reasons.append("checkpoint demo_usage missing")

    if bool(
        cfg.get("group_B_OOD_training_usage", True)
    ):
        reasons.append("group_B_OOD_training_usage is not False")

    if bool(
        cfg.get("group_B_OOD_checkpoint_usage", True)
    ):
        reasons.append("group_B_OOD_checkpoint_usage is not False")

    if not np.isclose(
        float(cfg.get("T", np.nan)),
        3.25,
        atol=1e-12,
    ):
        reasons.append(
            f"T={cfg.get('T')}"
        )

    if not np.isclose(
        float(cfg.get("dt", np.nan)),
        SOLVER_DT_EXPECTED,
        atol=1e-12,
    ):
        reasons.append(
            f"dt={cfg.get('dt')}"
        )

    base_idx = np.asarray(
        cfg.get("BASE_TIME_IDX", []),
        dtype=np.int64,
    ).reshape(-1)

    ood_idx = np.asarray(
        cfg.get("OOD_TIME_IDX", []),
        dtype=np.int64,
    ).reshape(-1)

    if not np.array_equal(
        base_idx,
        np.arange(0, 100, dtype=np.int64),
    ):
        reasons.append("BASE_TIME_IDX is not 0..99")

    if not np.array_equal(
        ood_idx,
        np.arange(100, 150, dtype=np.int64),
    ):
        reasons.append("OOD_TIME_IDX is not 100..149")

    run_tag = str(
        cfg.get("RUN_TAG", "")
    )

    if "strictDemo" not in run_tag:
        reasons.append(
            f"RUN_TAG lacks strictDemo: {run_tag}"
        )

    if "T3p25" not in run_tag:
        reasons.append(
            f"RUN_TAG lacks T3p25: {run_tag}"
        )

    return reasons


def locate_best_checkpoint(
    root: Path,
    expected_mode: str,
):
    if not root.is_dir():
        raise FileNotFoundError(
            f"Checkpoint root does not exist:\n{root}"
        )

    candidates = sorted(
        root.rglob("*best_trainmonitor_fullstate.pt")
    )

    if len(candidates) == 0:
        raise FileNotFoundError(
            "No *best_trainmonitor_fullstate.pt found under:\n"
            f"{root}"
        )

    valid = []
    rejected = []

    for p in candidates:
        try:
            ckpt = torch_load_any(
                p,
                map_location="cpu",
            )

            reasons = checkpoint_semantic_reasons(
                ckpt,
                expected_mode,
            )

            if len(reasons) == 0:
                valid.append(
                    (p, ckpt)
                )
            else:
                rejected.append(
                    (p, reasons)
                )

        except Exception as e:
            rejected.append(
                (p, [repr(e)])
            )

    if len(valid) != 1:
        lines = [
            f"Expected exactly one valid {expected_mode} strict-demo T3p25 BEST checkpoint.",
            "",
            "VALID candidates:",
        ]

        if valid:
            lines += [
                f"  {p}"
                for p, _ in valid
            ]
        else:
            lines.append("  (none)")

        lines.append("")
        lines.append("Rejected BEST checkpoints:")

        for p, reasons in rejected:
            lines.append(f"  {p}")
            for r in reasons:
                lines.append(f"    - {r}")

        raise RuntimeError(
            "\n".join(lines)
        )

    return valid[0]


NOMIXED_CKPT_PATH, nomixed_ckpt_cpu = (
    locate_best_checkpoint(
        NOMIXED_ROOT,
        "NO_MIXED",
    )
)

MIXED_CKPT_PATH, mixed_ckpt_cpu = (
    locate_best_checkpoint(
        MIXED_ROOT,
        "MIXED",
    )
)

print("\nLocated files:")
print("  dataset :", DATASET_PATH)
print("  no-mixed:", NOMIXED_CKPT_PATH)
print("  mixed   :", MIXED_CKPT_PATH)
print(
    "  no best : epoch",
    nomixed_ckpt_cpu.get("best_epoch"),
    "score",
    nomixed_ckpt_cpu.get("best_score"),
)
print(
    "  mix best: epoch",
    mixed_ckpt_cpu.get("best_epoch"),
    "score",
    mixed_ckpt_cpu.get("best_score"),
)


# ============================================================
# 4. LOAD DATASET
# ============================================================

with np.load(
    DATASET_PATH,
    allow_pickle=False,
) as f:

    required = [
        "beta",
        "times",
        "x",
        "uc",
        "uf",
        "base_time_idx",
        "ood_time_idx",
        "group_a_op_idx",
        "group_b_op_idx",
        "T",
        "dt",
        "base_snapshot_dt",
        "ood_snapshot_dt",
        "nx_lf",
        "nx_hf",
        "nr",
        "n_traj",
        "n_time",
        "lf_solver_tag",
        "hf_solver_tag",
    ]

    for key in required:
        if key not in f:
            raise KeyError(
                f"Dataset missing required key {key}.\n"
                f"Available keys: {f.files}"
            )

    beta = f["beta"][:].astype(
        np.float64
    ).reshape(-1)

    times = f["times"][:].astype(
        np.float64
    ).reshape(-1)

    x_grid = f["x"][:].astype(
        np.float32
    ).reshape(-1)

    uc = f["uc"][:].astype(
        np.float32
    )

    uf = f["uf"][:].astype(
        np.float32
    )

    BASE_TIME_IDX = (
        f["base_time_idx"][:]
        .astype(np.int64)
        .reshape(-1)
    )

    OOD_TIME_IDX = (
        f["ood_time_idx"][:]
        .astype(np.int64)
        .reshape(-1)
    )

    GROUP_A_DATASET = (
        f["group_a_op_idx"][:]
        .astype(np.int64)
        .reshape(-1)
    )

    GROUP_B_DATASET = (
        f["group_b_op_idx"][:]
        .astype(np.int64)
        .reshape(-1)
    )

    T_SAVED = float(
        np.asarray(f["T"]).reshape(-1)[0]
    )

    DT_SAVED = float(
        np.asarray(f["dt"]).reshape(-1)[0]
    )

    BASE_SNAPSHOT_DT_SAVED = float(
        np.asarray(
            f["base_snapshot_dt"]
        ).reshape(-1)[0]
    )

    OOD_SNAPSHOT_DT_SAVED = float(
        np.asarray(
            f["ood_snapshot_dt"]
        ).reshape(-1)[0]
    )

    NX_LF_SAVED = int(
        np.asarray(f["nx_lf"]).reshape(-1)[0]
    )

    NX_HF_SAVED = int(
        np.asarray(f["nx_hf"]).reshape(-1)[0]
    )

    NR_SAVED = int(
        np.asarray(f["nr"]).reshape(-1)[0]
    )

    N_TRAJ_SAVED = int(
        np.asarray(f["n_traj"]).reshape(-1)[0]
    )

    N_TIME_SAVED = int(
        np.asarray(f["n_time"]).reshape(-1)[0]
    )

    LF_SOLVER_TAG = scalar_str(
        f["lf_solver_tag"]
    )

    HF_SOLVER_TAG = scalar_str(
        f["hf_solver_tag"]
    )


N_OP, N_TRAJ, N_TIME, NX = uc.shape

assert (
    N_OP,
    N_TRAJ,
    N_TIME,
    NX,
) == (
    N_OP_EXPECTED,
    N_TRAJ_EXPECTED,
    N_TIME_EXPECTED,
    NR_EXPECTED,
)

assert uf.shape == uc.shape
assert beta.shape == (N_OP,)
assert times.shape == (N_TIME,)
assert x_grid.shape == (NX,)

assert NR_SAVED == NR_EXPECTED
assert N_TRAJ_SAVED == N_TRAJ_EXPECTED
assert N_TIME_SAVED == N_TIME_EXPECTED
assert NX_LF_SAVED == 72
assert NX_HF_SAVED == 256

assert np.isclose(
    T_SAVED,
    3.25,
    atol=1e-12,
)

assert np.isclose(
    DT_SAVED,
    SOLVER_DT_EXPECTED,
    atol=1e-12,
)

assert np.isclose(
    BASE_SNAPSHOT_DT_SAVED,
    SNAPSHOT_DT_EXPECTED,
    atol=1e-12,
)

assert np.isclose(
    OOD_SNAPSHOT_DT_SAVED,
    SNAPSHOT_DT_EXPECTED,
    atol=1e-12,
)

assert np.array_equal(
    BASE_TIME_IDX,
    np.arange(0, 100, dtype=np.int64),
)

assert np.array_equal(
    OOD_TIME_IDX,
    np.arange(100, 150, dtype=np.int64),
)

assert np.isclose(
    times[BASE_TIME_IDX[0]],
    BASE_MIN_EXPECTED,
    atol=1e-12,
)

assert np.isclose(
    times[BASE_TIME_IDX[-1]],
    BASE_MAX_EXPECTED,
    atol=1e-12,
)

assert np.isclose(
    times[OOD_TIME_IDX[0]],
    OOD_MIN_EXPECTED,
    atol=1e-12,
)

assert np.isclose(
    times[OOD_TIME_IDX[-1]],
    OOD_MAX_EXPECTED,
    atol=1e-12,
)

assert np.allclose(
    np.diff(times),
    SNAPSHOT_DT_EXPECTED,
    atol=1e-12,
    rtol=0.0,
)

assert np.all(np.isfinite(uc))
assert np.all(np.isfinite(uf))

print("\nDataset:")
print("  operator definition : beta")
print("  uc / uf             :", uc.shape, uf.shape)
print(
    "  BASE                :",
    f"{times[BASE_TIME_IDX[0]]:.3f}..{times[BASE_TIME_IDX[-1]]:.3f}",
    f"({len(BASE_TIME_IDX)})",
)
print(
    "  OOD                 :",
    f"{times[OOD_TIME_IDX[0]]:.3f}..{times[OOD_TIME_IDX[-1]]:.3f}",
    f"({len(OOD_TIME_IDX)})",
)
print(
    "  solver / snapshot dt:",
    DT_SAVED,
    "/",
    BASE_SNAPSHOT_DT_SAVED,
)
print("  LF solver tag       :", LF_SOLVER_TAG)
print("  HF solver tag       :", HF_SOLVER_TAG)


# ============================================================
# 5. CHECKPOINT / DATASET / PROTOCOL HARD AUDIT
# ============================================================

cfg_no = nomixed_ckpt_cpu["config"]
cfg_mix = mixed_ckpt_cpu["config"]

group_a_no = np.asarray(
    cfg_no["GROUP_A_OPS"],
    dtype=np.int64,
)

group_b_no = np.asarray(
    cfg_no["GROUP_B_OPS"],
    dtype=np.int64,
)

group_a_mix = np.asarray(
    cfg_mix["GROUP_A_OPS"],
    dtype=np.int64,
)

group_b_mix = np.asarray(
    cfg_mix["GROUP_B_OPS"],
    dtype=np.int64,
)

assert np.array_equal(
    group_a_no,
    group_a_mix,
)

assert np.array_equal(
    group_b_no,
    group_b_mix,
)

assert np.array_equal(
    group_a_no,
    np.sort(GROUP_A_DATASET),
)

assert np.array_equal(
    group_b_no,
    np.sort(GROUP_B_DATASET),
)

assert group_a_no.shape == (
    N_GROUP_EXPECTED,
)

assert group_b_no.shape == (
    N_GROUP_EXPECTED,
)

assert np.intersect1d(
    group_a_no,
    group_b_no,
).size == 0

assert np.array_equal(
    np.sort(
        np.concatenate(
            [group_a_no, group_b_no]
        )
    ),
    np.arange(
        N_OP,
        dtype=np.int64,
    ),
)

# Independent seed-42 audit.
rng_split = np.random.default_rng(42)
perm = rng_split.permutation(N_OP)

expected_group_b = np.sort(
    perm[:100]
).astype(np.int64)

expected_group_a = np.sort(
    perm[100:]
).astype(np.int64)

assert np.array_equal(
    group_a_no,
    expected_group_a,
)

assert np.array_equal(
    group_b_no,
    expected_group_b,
)

GROUP_A_OPS = group_a_no
GROUP_B_OPS = group_b_no

base_no = np.asarray(
    cfg_no["BASE_TIME_IDX"],
    dtype=np.int64,
)

base_mix = np.asarray(
    cfg_mix["BASE_TIME_IDX"],
    dtype=np.int64,
)

ood_no = np.asarray(
    cfg_no["OOD_TIME_IDX"],
    dtype=np.int64,
)

ood_mix = np.asarray(
    cfg_mix["OOD_TIME_IDX"],
    dtype=np.int64,
)

assert np.array_equal(
    base_no,
    base_mix,
)

assert np.array_equal(
    ood_no,
    ood_mix,
)

assert np.array_equal(
    base_no,
    BASE_TIME_IDX,
)

assert np.array_equal(
    ood_no,
    OOD_TIME_IDX,
)

assert cfg_no["operator_parameter"] == "beta"
assert cfg_mix["operator_parameter"] == "beta"

assert cfg_no["operator_definition"] == cfg_mix["operator_definition"]

assert cfg_no["trajectory_identity"] == "fixed_(beta,u0)"
assert cfg_mix["trajectory_identity"] == "fixed_(beta,u0)"

assert bool(
    cfg_no["demo_usage_tracking"]
)
assert bool(
    cfg_mix["demo_usage_tracking"]
)

assert bool(
    cfg_no["group_B_OOD_training_usage"]
) is False

assert bool(
    cfg_mix["group_B_OOD_training_usage"]
) is False

assert bool(
    cfg_no["group_B_OOD_checkpoint_usage"]
) is False

assert bool(
    cfg_mix["group_B_OOD_checkpoint_usage"]
) is False

# Normalization must match exactly.
norm_no = nomixed_ckpt_cpu["norm"]
norm_mix = mixed_ckpt_cpu["norm"]

for key in [
    "x_mean",
    "x_std",
    "t_mean",
    "t_std",
    "uc_mean",
    "uc_std",
    "uf_mean",
    "uf_std",
]:
    a = float(norm_no[key])
    b = float(norm_mix[key])

    if abs(a - b) > 1.0e-12:
        raise RuntimeError(
            f"Normalization mismatch for {key}: "
            f"{a} vs {b}"
        )

# Architecture must be exactly matched and expected.
for key, expected in PREFERRED_ARCH.items():
    if key == "DROPOUT":
        assert np.isclose(
            float(cfg_no[key]),
            expected,
            atol=1e-12,
        )
        assert np.isclose(
            float(cfg_mix[key]),
            expected,
            atol=1e-12,
        )
    else:
        assert int(cfg_no[key]) == int(expected)
        assert int(cfg_mix[key]) == int(expected)

for key in [
    "NX",
    "MAX_CTX_ROWS",
    "D_MODEL",
    "N_HEAD",
    "N_LAYERS",
    "D_FF",
    "DROPOUT",
    "N_PARAMS",
]:
    if isinstance(
        cfg_no[key],
        (float, np.floating),
    ):
        assert np.isclose(
            float(cfg_no[key]),
            float(cfg_mix[key]),
            atol=1e-12,
        )
    else:
        assert cfg_no[key] == cfg_mix[key]

assert int(cfg_no["NX"]) == NX
assert int(cfg_no["MAX_CTX_ROWS"]) == MAX_CTX_ROWS

print("\nProtocol audit PASSED:")
print("  operator parameter              : beta")
print("  trajectories/operator           : 20")
print("  A/B split dataset = checkpoints : YES")
print("  A/B split seed-42               : YES")
print(
    "  BASE pool identical             :",
    f"{times[BASE_TIME_IDX[0]]:.3f}..{times[BASE_TIME_IDX[-1]]:.3f}",
    f"({len(BASE_TIME_IDX)} tokens)",
)
print(
    "  OOD pool identical              :",
    f"{times[OOD_TIME_IDX[0]]:.3f}..{times[OOD_TIME_IDX[-1]]:.3f}",
    f"({len(OOD_TIME_IDX)} tokens)",
)
print("  normalization identical         : YES")
print("  architecture identical          : YES")
print("  strict demo tracking            : YES")
print("  Group-B OOD in training         : NO")
print("  Group-B OOD for checkpoint      : NO")


# ============================================================
# 6. STRICT BEST-CHECKPOINT DEMO-USAGE AUDIT
# ============================================================

def extract_best_usage(ckpt, label):
    if "demo_usage" not in ckpt:
        raise RuntimeError(
            f"{label} BEST checkpoint lacks demo_usage"
        )

    d = ckpt["demo_usage"]

    if "demo_usage_count" not in d:
        raise RuntimeError(
            f"{label} BEST demo_usage lacks demo_usage_count"
        )

    usage = np.asarray(
        d["demo_usage_count"],
        dtype=np.int64,
    )

    if usage.shape != (
        N_OP,
        N_TRAJ,
        N_TIME,
    ):
        raise RuntimeError(
            f"{label} usage shape mismatch: "
            f"{usage.shape}"
        )

    if np.any(usage < 0):
        raise RuntimeError(
            f"{label} usage contains negative values"
        )

    return usage


usage_no = extract_best_usage(
    nomixed_ckpt_cpu,
    "NO-MIXED",
)

usage_mix = extract_best_usage(
    mixed_ckpt_cpu,
    "MIXED",
)

# Absolutely no Group-B OOD demonstration usage is allowed.
gb_ood_no = int(
    np.sum(
        usage_no[
            GROUP_B_OPS
        ][
            :,
            :,
            OOD_TIME_IDX,
        ]
    )
)

gb_ood_mix = int(
    np.sum(
        usage_mix[
            GROUP_B_OPS
        ][
            :,
            :,
            OOD_TIME_IDX,
        ]
    )
)

assert gb_ood_no == 0
assert gb_ood_mix == 0

# NO-MIXED must not use A OOD either.
ga_ood_no = int(
    np.sum(
        usage_no[
            GROUP_A_OPS
        ][
            :,
            :,
            OOD_TIME_IDX,
        ]
    )
)

assert ga_ood_no == 0

# Shared-used BASE pool for every Group-B (operator,trajectory).
shared_used_mask = (
    (usage_no > 0)
    &
    (usage_mix > 0)
)

shared_base_counts = np.zeros(
    (
        len(GROUP_B_OPS),
        N_TRAJ,
    ),
    dtype=np.int64,
)

no_base_counts = np.zeros_like(
    shared_base_counts
)

mix_base_counts = np.zeros_like(
    shared_base_counts
)

# IMPORTANT NUMPY INDEXING NOTE:
# Avoid usage[op, :, BASE_TIME_IDX] here. Because BASE_TIME_IDX is
# advanced indexing, NumPy moves that indexed axis to the front and
# returns shape [Nbase, Ntraj] instead of [Ntraj, Nbase].
#
# First select Group-B operators, then index the time axis explicitly.
# Shapes:
#   usage_no[GROUP_B_OPS]                  -> [100,20,150]
#   usage_no[GROUP_B_OPS][:,:,BASE_TIME_IDX] -> [100,20,100]
# Count along axis=2 (BASE time) -> [100,20].

no_base_counts = np.count_nonzero(
    usage_no[
        GROUP_B_OPS
    ][
        :,
        :,
        BASE_TIME_IDX,
    ] > 0,
    axis=2,
).astype(
    np.int64
)

mix_base_counts = np.count_nonzero(
    usage_mix[
        GROUP_B_OPS
    ][
        :,
        :,
        BASE_TIME_IDX,
    ] > 0,
    axis=2,
).astype(
    np.int64
)

shared_base_counts = np.count_nonzero(
    shared_used_mask[
        GROUP_B_OPS
    ][
        :,
        :,
        BASE_TIME_IDX,
    ],
    axis=2,
).astype(
    np.int64
)

assert no_base_counts.shape == (
    len(GROUP_B_OPS),
    N_TRAJ,
)

assert mix_base_counts.shape == (
    len(GROUP_B_OPS),
    N_TRAJ,
)

assert shared_base_counts.shape == (
    len(GROUP_B_OPS),
    N_TRAJ,
)

print("\nStrict BEST demo-usage coverage:")
print(
    "  NO Group-B used BASE slots/traj   min/mean/max:",
    int(no_base_counts.min()),
    f"{float(no_base_counts.mean()):.3f}",
    int(no_base_counts.max()),
)
print(
    "  MIX Group-B used BASE slots/traj  min/mean/max:",
    int(mix_base_counts.min()),
    f"{float(mix_base_counts.mean()):.3f}",
    int(mix_base_counts.max()),
)
print(
    "  SHARED used BASE slots/traj        min/mean/max:",
    int(shared_base_counts.min()),
    f"{float(shared_base_counts.mean()):.3f}",
    int(shared_base_counts.max()),
)
print("  NO Group-B OOD demo uses          :", gb_ood_no)
print("  MIX Group-B OOD demo uses         :", gb_ood_mix)
print("  NO Group-A OOD demo uses          :", ga_ood_no)
print(
    "  MIX Group-A OOD demo uses         :",
    int(
        np.sum(
            usage_mix[
                GROUP_A_OPS
            ][
                :,
                :,
                OOD_TIME_IDX,
            ]
        )
    ),
)

if np.any(
    shared_base_counts
    < FIXED_M
):
    bad = np.argwhere(
        shared_base_counts
        < FIXED_M
    )

    details = []

    for row in bad[:50]:
        op_local, traj = map(
            int,
            row,
        )
        op = int(
            GROUP_B_OPS[
                op_local
            ]
        )
        nshared = int(
            shared_base_counts[
                op_local,
                traj,
            ]
        )
        details.append(
            f"(op={op}, traj={traj}, shared={nshared})"
        )

    raise RuntimeError(
        "STRICT FORMAL EVAL INVALID: some Group-B "
        "(operator,trajectory) pairs have fewer than "
        f"{FIXED_M} BASE demonstrations actually used "
        "by BOTH selected BEST models.\n"
        + "\n".join(details)
    )

print(
    "  STRICT K=5 shared-demo validity   : PASS for all",
    len(GROUP_B_OPS) * N_TRAJ,
    "Group-B trajectories",
)


# ============================================================
# 7. SELECT EXACT SAME STRICT M=5 DEMOS FOR BOTH MODELS
# ============================================================

# One deterministic M=5 context per Group-B (operator,trajectory).
# The same context is reused across all 50 OOD query times so the
# time-wise curve isolates query horizon rather than context noise.

demo_rng = np.random.default_rng(
    EVAL_SEED
)

selected_demo_times = np.empty(
    (
        N_OP,
        N_TRAJ,
        FIXED_M,
    ),
    dtype=np.int64,
)

selected_demo_times.fill(
    -1
)

demo_rows = []

for op in GROUP_B_OPS:
    op = int(op)

    for traj in range(N_TRAJ):
        shared_pool = BASE_TIME_IDX[
            shared_used_mask[
                op,
                traj,
                BASE_TIME_IDX,
            ]
        ]

        if len(shared_pool) < FIXED_M:
            raise RuntimeError(
                f"Insufficient strict shared pool for "
                f"op={op}, traj={traj}: {len(shared_pool)}"
            )

        demos = demo_rng.choice(
            shared_pool,
            size=FIXED_M,
            replace=False,
        ).astype(np.int64)

        # Keep random sampled order, matching training semantics.
        selected_demo_times[
            op,
            traj,
            :,
        ] = demos

        # Absolute hard check: selected demo was used by both BEST models.
        assert np.all(
            usage_no[
                op,
                traj,
                demos,
            ] > 0
        )

        assert np.all(
            usage_mix[
                op,
                traj,
                demos,
            ] > 0
        )

        demo_rows.append({
            "operator":
                op,
            "beta":
                float(beta[op]),
            "trajectory":
                traj,
            "n_no_used_BASE":
                int(
                    np.count_nonzero(
                        usage_no[
                            op,
                            traj,
                            BASE_TIME_IDX,
                        ] > 0
                    )
                ),
            "n_mix_used_BASE":
                int(
                    np.count_nonzero(
                        usage_mix[
                            op,
                            traj,
                            BASE_TIME_IDX,
                        ] > 0
                    )
                ),
            "n_shared_used_BASE":
                int(
                    len(shared_pool)
                ),
            "demo_time_indices":
                ",".join(
                    str(int(v))
                    for v in demos
                ),
            "demo_times":
                ",".join(
                    f"{times[int(v)]:.3f}"
                    for v in demos
                ),
            "no_usage_counts_for_selected":
                ",".join(
                    str(int(v))
                    for v in usage_no[
                        op,
                        traj,
                        demos,
                    ]
                ),
            "mix_usage_counts_for_selected":
                ",".join(
                    str(int(v))
                    for v in usage_mix[
                        op,
                        traj,
                        demos,
                    ]
                ),
        })


STRICT_DEMOS_CSV = (
    OUT_DIR
    / "strict_shared_demo_selection.csv"
)

pd.DataFrame(
    demo_rows
).to_csv(
    STRICT_DEMOS_CSV,
    index=False,
)

# Compact NPZ of exact selected strict context.
STRICT_DEMOS_NPZ = (
    OUT_DIR
    / "strict_shared_demo_selection.npz"
)

np.savez_compressed(
    STRICT_DEMOS_NPZ,
    selected_demo_times=
        selected_demo_times,
    group_b_ops=
        GROUP_B_OPS,
    base_time_idx=
        BASE_TIME_IDX,
    ood_time_idx=
        OOD_TIME_IDX,
    times=
        times,
    eval_seed=
        np.asarray(
            [EVAL_SEED],
            dtype=np.int64,
        ),
    definition=
        np.asarray(
            "M=5 BASE demos selected from intersection of "
            "NO-MIXED BEST and MIXED BEST exact gradient-demo usage"
        ),
)


# ============================================================
# 8. BUILD ALL 100,000 QUERY IDENTITIES
# ============================================================

N_CASES = (
    len(GROUP_B_OPS)
    * N_TRAJ
    * len(OOD_TIME_IDX)
)

assert N_CASES == 100_000

case_op = np.empty(
    N_CASES,
    dtype=np.int64,
)

case_traj = np.empty(
    N_CASES,
    dtype=np.int64,
)

case_qtime = np.empty(
    N_CASES,
    dtype=np.int64,
)

cursor = 0

for op in GROUP_B_OPS:
    op = int(op)

    for traj in range(N_TRAJ):
        n = len(
            OOD_TIME_IDX
        )

        case_op[
            cursor:
            cursor + n
        ] = op

        case_traj[
            cursor:
            cursor + n
        ] = traj

        case_qtime[
            cursor:
            cursor + n
        ] = OOD_TIME_IDX

        cursor += n

assert cursor == N_CASES

assert np.all(
    np.isin(
        case_op,
        GROUP_B_OPS,
    )
)

assert np.all(
    np.isin(
        case_qtime,
        OOD_TIME_IDX,
    )
)

print("\nFormal strict paired query set:")
print("  total cases        :", N_CASES)
print("  Group-B operators  :", len(np.unique(case_op)))
print("  trajectories/op    :", N_TRAJ)
print("  OOD query times    :", len(np.unique(case_qtime)))
print(
    "  cases/operator     :",
    N_TRAJ * len(OOD_TIME_IDX),
)
print(
    "  cases/OOD time     :",
    len(GROUP_B_OPS) * N_TRAJ,
)
print("  demos/case         :", FIXED_M)
print(
    "  demo rule          : strict shared-used BASE demos only"
)


# ============================================================
# 9. NORMALIZATION
# ============================================================

x_mean = float(
    norm_no["x_mean"]
)

x_std = float(
    norm_no["x_std"]
)

t_mean = float(
    norm_no["t_mean"]
)

t_std = float(
    norm_no["t_std"]
)

uc_mean = float(
    norm_no["uc_mean"]
)

uc_std = float(
    norm_no["uc_std"]
)

uf_mean = float(
    norm_no["uf_mean"]
)

uf_std = float(
    norm_no["uf_std"]
)

x_n = (
    (x_grid - x_mean)
    / x_std
).astype(
    np.float32
)

time_scalar_n = (
    (times - t_mean)
    / t_std
).astype(
    np.float32
)

time_rows_n = np.broadcast_to(
    time_scalar_n[:, None],
    (
        N_TIME,
        NX,
    ),
).copy()


# ============================================================
# 10. MODEL -- IDENTICAL TO TRAINING
# ============================================================

class RowSequenceTransformer(
    nn.Module
):
    def __init__(
        self,
        row_width,
        max_ctx_rows=23,
        n_row_types=7,
        d_model=288,
        nhead=6,
        num_layers=5,
        dim_ff=576,
        dropout=0.10,
    ):
        super().__init__()

        self.max_ctx_rows = max_ctx_rows

        self.row_in = nn.Linear(
            row_width,
            d_model,
        )

        self.row_type_emb = nn.Embedding(
            n_row_types,
            d_model,
        )

        self.pos_emb = nn.Embedding(
            max_ctx_rows,
            d_model,
        )

        enc_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=nhead,
            dim_feedforward=dim_ff,
            dropout=dropout,
            activation=torch.tanh,
            batch_first=True,
            norm_first=True,
        )

        self.encoder = nn.TransformerEncoder(
            enc_layer,
            num_layers=num_layers,
        )

        self.row_out = nn.Sequential(
            nn.LayerNorm(
                d_model
            ),
            nn.Linear(
                d_model,
                d_model,
            ),
            nn.Tanh(),
            nn.Linear(
                d_model,
                row_width,
            ),
        )

        self.register_buffer(
            "pos_ids",
            torch.arange(
                max_ctx_rows,
                dtype=torch.long,
            ),
            persistent=False,
        )

    def forward(
        self,
        mat_ctx,
        row_type_arr,
        key_padding_mask=None,
    ):
        B, S, _ = mat_ctx.shape

        if S > self.max_ctx_rows:
            raise ValueError(
                f"S={S} exceeds max_ctx_rows={self.max_ctx_rows}"
            )

        h = self.row_in(
            mat_ctx
        )

        h = (
            h
            + self.row_type_emb(
                row_type_arr
            )
            + self.pos_emb(
                self.pos_ids[:S]
            )[None, :, :]
        )

        h = self.encoder(
            h,
            src_key_padding_mask=
                key_padding_mask,
        )

        return self.row_out(
            h
        )


def make_model_from_cfg(cfg):
    model = RowSequenceTransformer(
        row_width=NX,
        max_ctx_rows=int(
            cfg["MAX_CTX_ROWS"]
        ),
        n_row_types=N_ROW_TYPES,
        d_model=int(
            cfg["D_MODEL"]
        ),
        nhead=int(
            cfg["N_HEAD"]
        ),
        num_layers=int(
            cfg["N_LAYERS"]
        ),
        dim_ff=int(
            cfg["D_FF"]
        ),
        dropout=float(
            cfg["DROPOUT"]
        ),
    ).to(
        DEVICE
    )

    n_params = sum(
        p.numel()
        for p in model.parameters()
        if p.requires_grad
    )

    assert n_params == int(
        cfg["N_PARAMS"]
    )

    return model


model_no = make_model_from_cfg(
    cfg_no
)

model_mix = make_model_from_cfg(
    cfg_mix
)

model_no.load_state_dict(
    nomixed_ckpt_cpu["model"]
)

model_mix.load_state_dict(
    mixed_ckpt_cpu["model"]
)

model_no.eval()
model_mix.eval()

print("\nModels loaded:")
print(
    "  parameters          :",
    f"{sum(p.numel() for p in model_no.parameters() if p.requires_grad):,}",
)
print(
    "  no-mixed best epoch :",
    int(
        nomixed_ckpt_cpu.get(
            "best_epoch",
            -1,
        )
    ),
)
print(
    "  mixed best epoch    :",
    int(
        mixed_ckpt_cpu.get(
            "best_epoch",
            -1,
        )
    ),
)
print(
    "  no best monitor     :",
    float(
        nomixed_ckpt_cpu.get(
            "best_score",
            np.nan,
        )
    ),
)
print(
    "  mix best monitor    :",
    float(
        mixed_ckpt_cpu.get(
            "best_score",
            np.nan,
        )
    ),
)


# ============================================================
# 11. FIXED M=5 ROW TYPES
# ============================================================

def row_types_for_m(
    m,
):
    out = []

    for _ in range(m):
        out.extend([
            TYPE_X_DEMO,
            TYPE_T_DEMO,
            TYPE_UC_DEMO,
            TYPE_UF_DEMO,
        ])

    out.extend([
        TYPE_X_QUERY,
        TYPE_T_QUERY,
        TYPE_UC_QUERY,
    ])

    return np.asarray(
        out,
        dtype=np.int64,
    )


ROW_TYPES_M5 = row_types_for_m(
    FIXED_M
)

S_M5 = (
    4 * FIXED_M
    + 3
)

assert S_M5 == MAX_CTX_ROWS
assert len(ROW_TYPES_M5) == S_M5

QUERY_POS = (
    4 * FIXED_M
    + 2
)

assert QUERY_POS == S_M5 - 1


# ============================================================
# 12. PAIRED EVALUATION
# ============================================================

raw_rel_all = np.empty(
    N_CASES,
    dtype=np.float64,
)

no_rel_all = np.empty(
    N_CASES,
    dtype=np.float64,
)

mix_rel_all = np.empty(
    N_CASES,
    dtype=np.float64,
)

no_r2_all = np.empty(
    N_CASES,
    dtype=np.float64,
)

mix_r2_all = np.empty(
    N_CASES,
    dtype=np.float64,
)

if SAVE_PREDICTIONS:
    pred_no_all = np.empty(
        (
            N_CASES,
            NX,
        ),
        dtype=np.float32,
    )

    pred_mix_all = np.empty(
        (
            N_CASES,
            NX,
        ),
        dtype=np.float32,
    )

    target_all = np.empty(
        (
            N_CASES,
            NX,
        ),
        dtype=np.float32,
    )

    raw_uc_all = np.empty(
        (
            N_CASES,
            NX,
        ),
        dtype=np.float32,
    )


@torch.inference_mode()
def run_paired_eval():
    print("\nRunning strict paired evaluation...")
    print("  eval batch size    :", EVAL_BATCH_SIZE)
    print("  vectorized builder : YES")

    eval_t0 = time.time()
    last_report = eval_t0

    # Fixed M=5 row locations.
    demo_x_pos = np.asarray(
        [0, 4, 8, 12, 16],
        dtype=np.int64,
    )

    demo_t_pos = demo_x_pos + 1
    demo_uc_pos = demo_x_pos + 2
    demo_uf_pos = demo_x_pos + 3

    q_x_pos = 20
    q_t_pos = 21
    q_uc_pos = 22

    for batch_id, start in enumerate(
        range(
            0,
            N_CASES,
            EVAL_BATCH_SIZE,
        )
    ):
        stop = min(
            start + EVAL_BATCH_SIZE,
            N_CASES,
        )

        ops = case_op[
            start:
            stop
        ]

        trajs = case_traj[
            start:
            stop
        ]

        qtimes = case_qtime[
            start:
            stop
        ]

        B = int(
            stop - start
        )

        # --------------------------------------------------------
        # Strict M=5 demos for this batch:
        #   demos shape = [B,5]
        # --------------------------------------------------------
        demos = selected_demo_times[
            ops,
            trajs,
            :
        ]

        assert demos.shape == (
            B,
            FIXED_M,
        )

        # Re-audit that every selected demonstration really appeared
        # as a gradient-training demo in BOTH selected BEST models.
        no_used = usage_no[
            ops[:, None],
            trajs[:, None],
            demos,
        ]

        mix_used = usage_mix[
            ops[:, None],
            trajs[:, None],
            demos,
        ]

        if not np.all(
            no_used > 0
        ):
            raise RuntimeError(
                "Selected NO-MIXED formal demo has zero BEST usage."
            )

        if not np.all(
            mix_used > 0
        ):
            raise RuntimeError(
                "Selected MIXED formal demo has zero BEST usage."
            )

        # --------------------------------------------------------
        # Vectorized context construction
        # --------------------------------------------------------
        ctx = np.zeros(
            (
                B,
                MAX_CTX_ROWS,
                NX,
            ),
            dtype=np.float32,
        )

        # x rows
        ctx[
            :,
            demo_x_pos,
            :
        ] = x_n[
            None,
            None,
            :
        ]

        ctx[
            :,
            q_x_pos,
            :
        ] = x_n[
            None,
            :
        ]

        # t rows: [B,5,NX]
        ctx[
            :,
            demo_t_pos,
            :
        ] = time_rows_n[
            demos
        ]

        ctx[
            :,
            q_t_pos,
            :
        ] = time_rows_n[
            qtimes
        ]

        # Gather demo physical states in one shot.
        #
        # Advanced indexing shapes:
        #   demo_uc_phys, demo_uf_phys -> [B,5,NX]
        demo_uc_phys = uc[
            ops[:, None],
            trajs[:, None],
            demos,
            :
        ]

        demo_uf_phys = uf[
            ops[:, None],
            trajs[:, None],
            demos,
            :
        ]

        ctx[
            :,
            demo_uc_pos,
            :
        ] = (
            (
                demo_uc_phys
                - uc_mean
            )
            / uc_std
        ).astype(
            np.float32,
            copy=False,
        )

        ctx[
            :,
            demo_uf_pos,
            :
        ] = (
            (
                demo_uf_phys
                - uf_mean
            )
            / uf_std
        ).astype(
            np.float32,
            copy=False,
        )

        # Query LF and HF target, shape [B,NX].
        raw_uc_phys = uc[
            ops,
            trajs,
            qtimes,
            :
        ]

        target_phys = uf[
            ops,
            trajs,
            qtimes,
            :
        ]

        ctx[
            :,
            q_uc_pos,
            :
        ] = (
            (
                raw_uc_phys
                - uc_mean
            )
            / uc_std
        ).astype(
            np.float32,
            copy=False,
        )

        # Fixed M=5 has no actual padding, but keep the same mask
        # semantics as training.
        row_types = np.broadcast_to(
            ROW_TYPES_M5[
                None,
                :
            ],
            (
                B,
                MAX_CTX_ROWS,
            ),
        )

        padding_mask = np.zeros(
            (
                B,
                MAX_CTX_ROWS,
            ),
            dtype=np.bool_,
        )

        ctx_t = torch.from_numpy(
            ctx
        ).to(
            DEVICE,
            non_blocking=True,
        )

        row_types_t = torch.from_numpy(
            np.asarray(
                row_types,
                dtype=np.int64,
            )
        ).to(
            DEVICE,
            non_blocking=True,
        )

        padding_mask_t = torch.from_numpy(
            padding_mask
        ).to(
            DEVICE,
            non_blocking=True,
        )

        # --------------------------------------------------------
        # Paired model inference
        # --------------------------------------------------------
        out_no_n = model_no(
            ctx_t,
            row_types_t,
            key_padding_mask=
                padding_mask_t,
        )[
            :,
            QUERY_POS,
            :
        ]

        out_mix_n = model_mix(
            ctx_t,
            row_types_t,
            key_padding_mask=
                padding_mask_t,
        )[
            :,
            QUERY_POS,
            :
        ]

        pred_no = (
            out_no_n
            * uf_std
            + uf_mean
        ).detach().cpu().numpy()

        pred_mix = (
            out_mix_n
            * uf_std
            + uf_mean
        ).detach().cpu().numpy()

        # --------------------------------------------------------
        # Metrics
        # --------------------------------------------------------
        raw_rel = rel_l2_np(
            raw_uc_phys,
            target_phys,
        )

        no_rel = rel_l2_np(
            pred_no,
            target_phys,
        )

        mix_rel = rel_l2_np(
            pred_mix,
            target_phys,
        )

        no_r2 = r2_np(
            pred_no,
            target_phys,
        )

        mix_r2 = r2_np(
            pred_mix,
            target_phys,
        )

        raw_rel_all[
            start:
            stop
        ] = raw_rel

        no_rel_all[
            start:
            stop
        ] = no_rel

        mix_rel_all[
            start:
            stop
        ] = mix_rel

        no_r2_all[
            start:
            stop
        ] = no_r2

        mix_r2_all[
            start:
            stop
        ] = mix_r2

        if SAVE_PREDICTIONS:
            pred_no_all[
                start:
                stop
            ] = pred_no

            pred_mix_all[
                start:
                stop
            ] = pred_mix

            target_all[
                start:
                stop
            ] = target_phys

            raw_uc_all[
                start:
                stop
            ] = raw_uc_phys

        # --------------------------------------------------------
        # Frequent progress reporting
        # --------------------------------------------------------
        now = time.time()

        if (
            batch_id == 0
            or stop == N_CASES
            or (
                now - last_report
                >= 10.0
            )
        ):
            elapsed = max(
                now - eval_t0,
                1.0e-9,
            )

            rate = (
                stop
                / elapsed
            )

            print(
                f"  {stop:>7d}/{N_CASES} cases "
                f"| {rate:8.1f} cases/s"
            )

            last_report = now

        # Release GPU references promptly.
        del (
            ctx_t,
            row_types_t,
            padding_mask_t,
            out_no_n,
            out_mix_n,
        )


run_paired_eval()


# ============================================================
# 13. GLOBAL METRICS
# ============================================================

raw_summary = summarize(
    raw_rel_all,
    scale=100.0,
)

no_summary = summarize(
    no_rel_all,
    scale=100.0,
)

mix_summary = summarize(
    mix_rel_all,
    scale=100.0,
)

no_r2_summary = summarize(
    no_r2_all
)

mix_r2_summary = summarize(
    mix_r2_all
)

no_vs_raw_reduction = (
    1.0
    - no_summary["mean"]
    / raw_summary["mean"]
) * 100.0

mix_vs_raw_reduction = (
    1.0
    - mix_summary["mean"]
    / raw_summary["mean"]
) * 100.0

mix_vs_no_reduction = (
    1.0
    - mix_summary["mean"]
    / no_summary["mean"]
) * 100.0

paired_gain_pp = (
    100.0
    * (
        no_rel_all
        - mix_rel_all
    )
)

paired_win = (
    mix_rel_all
    < no_rel_all
)

paired_tie = np.isclose(
    mix_rel_all,
    no_rel_all,
    rtol=0.0,
    atol=1.0e-12,
)

paired_win_rate = (
    100.0
    * float(
        np.mean(
            paired_win
        )
    )
)

paired_tie_rate = (
    100.0
    * float(
        np.mean(
            paired_tie
        )
    )
)

print("\n" + "=" * 136)
print("FORMAL STRICT GROUP-B OOD RESULTS -- ALL 100,000 QUERY STATES")
print("=" * 136)
print(
    "RAW LF     :",
    summary_str(
        raw_summary
    ),
)
print(
    "NO-MIXED   :",
    summary_str(
        no_summary
    ),
)
print(
    "MIXED      :",
    summary_str(
        mix_summary
    ),
)
print("-")
print(
    f"NO-MIXED reduction vs RAW : "
    f"{no_vs_raw_reduction:.3f}%"
)
print(
    f"MIXED reduction vs RAW    : "
    f"{mix_vs_raw_reduction:.3f}%"
)
print(
    f"MIXED reduction vs NO     : "
    f"{mix_vs_no_reduction:.3f}%"
)
print(
    f"MIXED paired win rate     : "
    f"{paired_win_rate:.3f}%"
)
print(
    f"MIXED paired tie rate     : "
    f"{paired_tie_rate:.3f}%"
)
print(
    "Paired mean gain (NO - MIX), "
    "RelL2 percentage points   :",
    f"{float(np.mean(paired_gain_pp)):.4f}",
)
print(
    "Paired median gain (NO - MIX), "
    "RelL2 percentage points :",
    f"{float(np.median(paired_gain_pp)):.4f}",
)
print("-")
print(
    "NO-MIXED R2 mean  :",
    f"{no_r2_summary['mean']:.6f}",
)
print(
    "MIXED R2 mean     :",
    f"{mix_r2_summary['mean']:.6f}",
)


# ============================================================
# 14. PER-TIME SUMMARY
# ============================================================

time_rows = []

for qti in OOD_TIME_IDX:
    mask = (
        case_qtime
        == qti
    )

    raw_t = (
        100.0
        * raw_rel_all[
            mask
        ]
    )

    no_t = (
        100.0
        * no_rel_all[
            mask
        ]
    )

    mix_t = (
        100.0
        * mix_rel_all[
            mask
        ]
    )

    mean_no = float(
        np.mean(
            no_t
        )
    )

    mean_mix = float(
        np.mean(
            mix_t
        )
    )

    reduction = (
        (
            1.0
            - mean_mix
            / mean_no
        )
        * 100.0
        if mean_no > 0.0
        else np.nan
    )

    win_rate = (
        100.0
        * float(
            np.mean(
                mix_t
                < no_t
            )
        )
    )

    time_rows.append({
        "query_time_idx":
            int(qti),
        "query_time":
            float(times[qti]),
        "n_cases":
            int(np.sum(mask)),
        "raw_mean_relL2_pct":
            float(
                np.mean(
                    raw_t
                )
            ),
        "nomixed_mean_relL2_pct":
            mean_no,
        "mixed_mean_relL2_pct":
            mean_mix,
        "mixed_vs_nomixed_mean_reduction_pct":
            float(
                reduction
            ),
        "mixed_paired_win_rate_pct":
            float(
                win_rate
            ),
        "nomixed_R2_mean":
            float(
                np.mean(
                    no_r2_all[
                        mask
                    ]
                )
            ),
        "mixed_R2_mean":
            float(
                np.mean(
                    mix_r2_all[
                        mask
                    ]
                )
            ),
    })


by_time_df = pd.DataFrame(
    time_rows
)

BY_TIME_CSV = (
    OUT_DIR
    / "by_time.csv"
)

by_time_df.to_csv(
    BY_TIME_CSV,
    index=False,
)

print("\nPer-time mean RelL2:")
print(
    by_time_df[
        [
            "query_time",
            "raw_mean_relL2_pct",
            "nomixed_mean_relL2_pct",
            "mixed_mean_relL2_pct",
            "mixed_vs_nomixed_mean_reduction_pct",
            "mixed_paired_win_rate_pct",
        ]
    ].to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}",
    )
)


# ============================================================
# 15. PER-OPERATOR SUMMARY
# ============================================================

op_rows = []

for op in GROUP_B_OPS:
    op = int(op)

    mask = (
        case_op
        == op
    )

    raw_o = (
        100.0
        * raw_rel_all[
            mask
        ]
    )

    no_o = (
        100.0
        * no_rel_all[
            mask
        ]
    )

    mix_o = (
        100.0
        * mix_rel_all[
            mask
        ]
    )

    mean_no = float(
        np.mean(
            no_o
        )
    )

    mean_mix = float(
        np.mean(
            mix_o
        )
    )

    op_rows.append({
        "operator":
            op,
        "beta":
            float(beta[op]),
        "n_cases":
            int(np.sum(mask)),
        "raw_mean_relL2_pct":
            float(
                np.mean(
                    raw_o
                )
            ),
        "nomixed_mean_relL2_pct":
            mean_no,
        "mixed_mean_relL2_pct":
            mean_mix,
        "mixed_vs_nomixed_mean_reduction_pct":
            float(
                (
                    1.0
                    - mean_mix
                    / mean_no
                )
                * 100.0
            ),
        "mixed_paired_win_rate_pct":
            float(
                100.0
                * np.mean(
                    mix_o
                    < no_o
                )
            ),
        "nomixed_R2_mean":
            float(
                np.mean(
                    no_r2_all[
                        mask
                    ]
                )
            ),
        "mixed_R2_mean":
            float(
                np.mean(
                    mix_r2_all[
                        mask
                    ]
                )
            ),
    })


by_operator_df = pd.DataFrame(
    op_rows
)

BY_OPERATOR_CSV = (
    OUT_DIR
    / "by_operator.csv"
)

by_operator_df.to_csv(
    BY_OPERATOR_CSV,
    index=False,
)


# ============================================================
# 16. CASE-LEVEL TABLE
# ============================================================

demo_index_strings = []
demo_time_strings = []

for op, traj in zip(
    case_op,
    case_traj,
):
    demos = selected_demo_times[
        int(op),
        int(traj),
    ]

    demo_index_strings.append(
        ",".join(
            str(int(v))
            for v in demos
        )
    )

    demo_time_strings.append(
        ",".join(
            f"{times[int(v)]:.3f}"
            for v in demos
        )
    )


case_df = pd.DataFrame({
    "case_id":
        np.arange(
            N_CASES,
            dtype=np.int64,
        ),
    "operator":
        case_op,
    "beta":
        beta[
            case_op
        ],
    "trajectory":
        case_traj,
    "demo_time_indices":
        demo_index_strings,
    "demo_times":
        demo_time_strings,
    "query_time_idx":
        case_qtime,
    "query_time":
        times[
            case_qtime
        ],
    "raw_relL2_pct":
        100.0
        * raw_rel_all,
    "nomixed_relL2_pct":
        100.0
        * no_rel_all,
    "mixed_relL2_pct":
        100.0
        * mix_rel_all,
    "paired_gain_no_minus_mix_pp":
        paired_gain_pp,
    "mixed_wins":
        paired_win.astype(
            np.int8
        ),
    "nomixed_R2":
        no_r2_all,
    "mixed_R2":
        mix_r2_all,
})

CASE_METRICS_CSV = (
    OUT_DIR
    / "case_metrics.csv"
)

case_df.to_csv(
    CASE_METRICS_CSV,
    index=False,
)


# ============================================================
# 17. EXACT EVALUATION CASE NPZ
# ============================================================

EXACT_CASES_NPZ = (
    OUT_DIR
    / "exact_eval_cases.npz"
)

case_demo_times = np.stack(
    [
        selected_demo_times[
            int(op),
            int(traj),
        ]
        for op, traj in zip(
            case_op,
            case_traj,
        )
    ],
    axis=0,
).astype(
    np.int64
)

np.savez_compressed(
    EXACT_CASES_NPZ,
    operator=
        case_op,
    beta=
        beta[
            case_op
        ],
    trajectory=
        case_traj,
    demo_time_idx=
        case_demo_times,
    demo_times=
        times[
            case_demo_times
        ],
    query_time_idx=
        case_qtime,
    query_time=
        times[
            case_qtime
        ],
    group_b_ops=
        GROUP_B_OPS,
    base_time_idx=
        BASE_TIME_IDX,
    ood_time_idx=
        OOD_TIME_IDX,
    eval_seed=
        np.asarray(
            [EVAL_SEED],
            dtype=np.int64,
        ),
    strict_definition=
        np.asarray(
            "each demo time has positive exact gradient-demo usage "
            "in BOTH selected BEST checkpoints"
        ),
)


# ============================================================
# 18. SAVE PREDICTIONS
# ============================================================

PREDICTIONS_NPZ = (
    OUT_DIR
    / "predictions.npz"
)

if SAVE_PREDICTIONS:
    np.savez_compressed(
        PREDICTIONS_NPZ,
        pred_nomixed=
            pred_no_all,
        pred_mixed=
            pred_mix_all,
        target=
            target_all,
        raw_uc=
            raw_uc_all,
        raw_relL2=
            raw_rel_all,
        nomixed_relL2=
            no_rel_all,
        mixed_relL2=
            mix_rel_all,
        nomixed_R2=
            no_r2_all,
        mixed_R2=
            mix_r2_all,
        operator=
            case_op,
        trajectory=
            case_traj,
        query_time_idx=
            case_qtime,
        query_time=
            times[
                case_qtime
            ],
        demo_time_idx=
            case_demo_times,
    )


# ============================================================
# 19. STRICT DEMO AUDIT JSON
# ============================================================

DEMO_AUDIT_JSON = (
    OUT_DIR
    / "strict_demo_usage_audit.json"
)

demo_audit_payload = {
    "definition": (
        "formal M=5 BASE demonstrations are selected only from "
        "the intersection of exact gradient-training demo usage "
        "recorded in the selected NO-MIXED BEST checkpoint and "
        "selected MIXED BEST checkpoint"
    ),
    "nomixed_best_checkpoint":
        str(
            NOMIXED_CKPT_PATH
        ),
    "mixed_best_checkpoint":
        str(
            MIXED_CKPT_PATH
        ),
    "nomixed_best_epoch":
        int(
            nomixed_ckpt_cpu.get(
                "best_epoch",
                -1,
            )
        ),
    "mixed_best_epoch":
        int(
            mixed_ckpt_cpu.get(
                "best_epoch",
                -1,
            )
        ),
    "group_B_ood_demo_uses_nomixed":
        gb_ood_no,
    "group_B_ood_demo_uses_mixed":
        gb_ood_mix,
    "group_A_ood_demo_uses_nomixed":
        ga_ood_no,
    "group_A_ood_demo_uses_mixed":
        int(
            np.sum(
                usage_mix[
                    GROUP_A_OPS
                ][
                    :,
                    :,
                    OOD_TIME_IDX,
                ]
            )
        ),
    "nomixed_group_B_used_BASE_slots_per_trajectory": {
        "min":
            int(
                no_base_counts.min()
            ),
        "mean":
            float(
                no_base_counts.mean()
            ),
        "max":
            int(
                no_base_counts.max()
            ),
    },
    "mixed_group_B_used_BASE_slots_per_trajectory": {
        "min":
            int(
                mix_base_counts.min()
            ),
        "mean":
            float(
                mix_base_counts.mean()
            ),
        "max":
            int(
                mix_base_counts.max()
            ),
    },
    "shared_group_B_used_BASE_slots_per_trajectory": {
        "min":
            int(
                shared_base_counts.min()
            ),
        "mean":
            float(
                shared_base_counts.mean()
            ),
        "max":
            int(
                shared_base_counts.max()
            ),
    },
    "strict_K5_valid_for_all_group_B_trajectories":
        bool(
            np.all(
                shared_base_counts
                >= FIXED_M
            )
        ),
    "n_group_B_trajectory_identities":
        int(
            len(GROUP_B_OPS)
            * N_TRAJ
        ),
}

with open(
    DEMO_AUDIT_JSON,
    "w",
) as f:
    json.dump(
        demo_audit_payload,
        f,
        indent=2,
    )


# ============================================================
# 20. SUMMARY JSON
# ============================================================

SUMMARY_JSON = (
    OUT_DIR
    / "summary.json"
)

summary_payload = {
    "protocol": {
        "dataset":
            str(
                DATASET_PATH
            ),
        "nomixed_best_checkpoint":
            str(
                NOMIXED_CKPT_PATH
            ),
        "mixed_best_checkpoint":
            str(
                MIXED_CKPT_PATH
            ),
        "operator_definition":
            "beta",
        "trajectories_per_operator":
            N_TRAJ,
        "group_B_operators":
            len(
                GROUP_B_OPS
            ),
        "base_range": [
            float(
                times[
                    BASE_TIME_IDX[
                        0
                    ]
                ]
            ),
            float(
                times[
                    BASE_TIME_IDX[
                        -1
                    ]
                ]
            ),
        ],
        "ood_range": [
            float(
                times[
                    OOD_TIME_IDX[
                        0
                    ]
                ]
            ),
            float(
                times[
                    OOD_TIME_IDX[
                        -1
                    ]
                ]
            ),
        ],
        "M":
            FIXED_M,
        "n_cases":
            N_CASES,
        "strict_demo_rule": (
            "selected from BASE-time intersection of positive "
            "BEST-checkpoint gradient-demo usage in both models"
        ),
        "same_context_for_both_models":
            True,
        "same_demo_context_reused_across_all_50_ood_queries_of_each_trajectory":
            True,
    },
    "raw_LF_RelL2_percent":
        raw_summary,
    "nomixed_RelL2_percent":
        no_summary,
    "mixed_RelL2_percent":
        mix_summary,
    "nomixed_R2":
        no_r2_summary,
    "mixed_R2":
        mix_r2_summary,
    "comparisons": {
        "nomixed_reduction_vs_raw_percent":
            float(
                no_vs_raw_reduction
            ),
        "mixed_reduction_vs_raw_percent":
            float(
                mix_vs_raw_reduction
            ),
        "mixed_reduction_vs_nomixed_percent":
            float(
                mix_vs_no_reduction
            ),
        "mixed_paired_win_rate_percent":
            float(
                paired_win_rate
            ),
        "mixed_paired_tie_rate_percent":
            float(
                paired_tie_rate
            ),
        "paired_mean_gain_no_minus_mix_percentage_points":
            float(
                np.mean(
                    paired_gain_pp
                )
            ),
        "paired_median_gain_no_minus_mix_percentage_points":
            float(
                np.median(
                    paired_gain_pp
                )
            ),
    },
    "strict_demo_audit":
        demo_audit_payload,
}

with open(
    SUMMARY_JSON,
    "w",
) as f:
    json.dump(
        summary_payload,
        f,
        indent=2,
    )


# ============================================================
# 21. PLOTS
# ============================================================

REL_TIME_PNG = (
    OUT_DIR
    / "relL2_vs_time.png"
)

plt.figure(
    figsize=(8.0, 5.2)
)

plt.plot(
    by_time_df[
        "query_time"
    ],
    by_time_df[
        "raw_mean_relL2_pct"
    ],
    marker="o",
    label="Raw LF",
)

plt.plot(
    by_time_df[
        "query_time"
    ],
    by_time_df[
        "nomixed_mean_relL2_pct"
    ],
    marker="o",
    label="NO-MIXED",
)

plt.plot(
    by_time_df[
        "query_time"
    ],
    by_time_df[
        "mixed_mean_relL2_pct"
    ],
    marker="o",
    label="MIXED",
)

plt.xlabel(
    "OOD query time"
)

plt.ylabel(
    "Mean RelL2 (%)"
)

plt.title(
    "KS Group-B temporal OOD: strict shared training demos"
)

plt.grid(
    alpha=0.25
)

plt.legend()

plt.tight_layout()

plt.savefig(
    REL_TIME_PNG,
    dpi=220,
    bbox_inches="tight",
)

plt.show()
plt.close()


PAIRED_PNG = (
    OUT_DIR
    / "paired_no_vs_mixed.png"
)

plt.figure(
    figsize=(6.2, 6.0)
)

plt.scatter(
    100.0
    * no_rel_all,
    100.0
    * mix_rel_all,
    s=7,
    alpha=0.18,
)

lim = float(
    max(
        np.max(
            100.0
            * no_rel_all
        ),
        np.max(
            100.0
            * mix_rel_all
        ),
    )
)

plt.plot(
    [0.0, lim],
    [0.0, lim],
    linestyle="--",
    linewidth=1.0,
)

plt.xlabel(
    "NO-MIXED RelL2 (%)"
)

plt.ylabel(
    "MIXED RelL2 (%)"
)

plt.title(
    "Paired Group-B OOD cases"
)

plt.grid(
    alpha=0.20
)

plt.tight_layout()

plt.savefig(
    PAIRED_PNG,
    dpi=220,
    bbox_inches="tight",
)

plt.show()
plt.close()


# ============================================================
# 22. FINAL OUTPUT LIST
# ============================================================

print("\nSaved outputs:")

saved_paths = [
    STRICT_DEMOS_CSV,
    STRICT_DEMOS_NPZ,
    BY_OPERATOR_CSV,
    BY_TIME_CSV,
    CASE_METRICS_CSV,
    EXACT_CASES_NPZ,
    DEMO_AUDIT_JSON,
    SUMMARY_JSON,
    REL_TIME_PNG,
    PAIRED_PNG,
]

if SAVE_PREDICTIONS:
    saved_paths.append(
        PREDICTIONS_NPZ
    )

for p in saved_paths:
    print(" ", p)

print("\n" + "=" * 136)
print("DONE")
print("=" * 136)
```
